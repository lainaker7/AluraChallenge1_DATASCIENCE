# 1. Análisis de la Facturación

```python
# Se empieza a revisar como han sido las ventas según los precios de los artículos que vende cada tienda
# para definimos la variable de cada tienda y sumamos sus ingresos de acuerdo a los precios.


ventas1 = tienda['Precio'].sum()
ventas2 = tienda2['Precio'].sum()
ventas3 = tienda3['Precio'].sum()
ventas4 = tienda4['Precio'].sum()


```

```python
print(f"💵 Tienda 1: ${ventas1:,.0f} pesos")
print(f"💵 Tienda 2: ${ventas2:,.0f} pesos")
print(f"💵 Tienda 3: ${ventas3:,.0f} pesos")
print(f"💵 Tienda 4: ${ventas4:,.0f} pesos")
```

```python
#Ahora hacemos una lista de todas esas ventas para poder comparar los mejores y peores ingresos de cada tienda.
ventas_totales = [ventas1, ventas2, ventas3, ventas4]
mejor_venta = max(ventas_totales)
peor_venta = min(ventas_totales)

 #Extraemos de los datos cual fue el mejor de los ingresos.
mejor_tienda = ventas_totales.index(mejor_venta) + 1
peor_tienda = ventas_totales.index(peor_venta) + 1
print(f'\n💰 Mejor venta: Tienda {mejor_tienda} tuvo el mayor ingreso')
print(f'⚠️ Peor venta: Tienda {peor_tienda} tuvo el menor ingreso')
```

```python
#Ahora buscamos graficar el resultado para tener la comparativa según el primer analisis realizado.
import matplotlib.pyplot as plt
```

```python

# Datos
ingresos = [ventas1, ventas2, ventas3, ventas4]
tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']
colores = ['#8BB657', '#337B9F', '#9AD1D0', '#B24223']

plt.figure(figsize=(10, 8))
plt.pie(ingresos,
        labels=tiendas,
        colors=colores,
        autopct='%1.1f%%',
        startangle=90,
        explode=(0.05, 0.05, 0.05, 0.05))  # Separar las rebanadas

plt.title('🚀 Ventas por tienda', fontsize=16, fontweight='bold')
plt.axis('equal')  # Para que sea un círculo perfecto
plt.show()

 # 💾 Guardar el gráfico como imagen PNG
plt.tight_layout()
plt.savefig('grafico1_ingresos_tiendas.png', dpi=300, bbox_inches='tight')

# Del análisis se puede verificar que de las ventas totales la tienda 1 tiene el mayor ingreso siendo el 26.1% del total mientras
# que la tienda 4 solo representa un 23.6% de las ventas globales.
```

# 2. Ventas por categoría

```python
# En esta parte analizaremos cuales son los productos más vendidos de cada tienda, para descubrir el potencial de cada tienda
# esto también permite ver la rotación de stock y ver que productos se pueden sacar de la línea de producción.

print('Listado de los productos más vendidos en cada cada tienda')
print("\nLos productos más comprados en la Tienda 1 son :")
productos_1 = tienda['Categoría del Producto'].value_counts().head(5)
print(productos_1)

```

```python
print('Listado de los productos más vendidos en cada cada tienda')
print("\nLos productos más comprados en la Tienda 2 son :")
productos_2 = tienda2['Categoría del Producto'].value_counts().head(5)
print(productos_2)
```

```python
print('Listado de los productos más vendidos en cada cada tienda')
print("\nLos productos más comprados en la Tienda 3 son :")
productos_3 = tienda3['Categoría del Producto'].value_counts().head(5)
print(productos_3)
```

```python
print('Listado de los productos más vendidos en cada cada tienda')
print("\nLos productos más comprados en la Tienda 4 son :")
productos_4 = tienda4['Categoría del Producto'].value_counts().head(5)
print(productos_4)
```

```python
# Combinar datos de todas las tiendas para comparación
df_completo = pd.concat([
    tienda.assign(Tienda='Tienda 1'),
    tienda2.assign(Tienda='Tienda 2'),
    tienda3.assign(Tienda='Tienda 3'),
    tienda4.assign(Tienda='Tienda 4')
])

# Obtener top categorías generales
top_categorias_general = df_completo['Categoría del Producto'].value_counts().head(5).index

# Crear pivot table para comparación
pivot_ventas = df_completo[df_completo['Categoría del Producto'].isin(top_categorias_general)].pivot_table(
    index='Categoría del Producto',
    columns='Tienda',
    values='Precio',
    aggfunc='sum',
    fill_value=0
)

# Gráfica de barras agrupadas
plt.figure(figsize=(14, 8))
pivot_ventas.plot(kind='bar', figsize=(14, 8), width=0.8)
plt.title('📊 COMPARACIÓN: Ventas por Categoría en Todas las Tiendas', fontsize=16, fontweight='bold')
plt.xlabel('Categoría del Producto')
plt.ylabel('Total de Ventas')
plt.legend(title='Tienda')
plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

###** En el gráfico anterior se puede visualizar que todas las tiendas las categorías más vendidas, son los electrodomésticos y los electrónicos.**

# 3. Calificación promedio de la tienda


```python
# En esta parte veremos el nivel de satisfacción de cada uno de los clientes al momento de su compra.

import seaborn as sns
import numpy as np
```

```python
## Análisis estadístico detallado
print("🏆 COMPARATIVA DE CALIFICACIONES POR TIENDA")
print("="*55)

# Crear lista de tiendas corregida
tiendas_data = [
    ('Tienda 1', tienda),
    ('Tienda 2', tienda2),
    ('Tienda 3', tienda3),
    ('Tienda 4', tienda4)
]

resultados = []

for nombre, df in tiendas_data:
    calificaciones = df['Calificación']
    resultados.append({
        'Tienda': nombre,
        'Promedio': calificaciones.mean(),
        'Mediana': calificaciones.median(),
        'Desviación': calificaciones.std(),
        'Mínimo': calificaciones.min(),
        'Máximo': calificaciones.max(),
        'Total': len(calificaciones),
        '5 estrellas': (calificaciones == 5).sum(),
        '% 5 estrellas': ((calificaciones == 5).sum() / len(calificaciones)) * 100
    })

df_comparativo = pd.DataFrame(resultados)

# Identificar mejor y peor tienda
mejor_tienda = df_comparativo.loc[df_comparativo['Promedio'].idxmax()]
peor_tienda = df_comparativo.loc[df_comparativo['Promedio'].idxmin()]

print(f"\n🎯 MEJOR TIENDA: {mejor_tienda['Tienda']}")
print(f"   Calificación promedio: {mejor_tienda['Promedio']:.2f}")
print(f"   Porcentaje de 5 estrellas: {mejor_tienda['% 5 estrellas']:.1f}%")

print(f"\n⚠️  PEOR TIENDA: {peor_tienda['Tienda']}")
print(f"   Calificación promedio: {peor_tienda['Promedio']:.2f}")
print(f"   Porcentaje de 5 estrellas: {peor_tienda['% 5 estrellas']:.1f}%")

# Diferencia entre mejor y peor
diferencia = mejor_tienda['Promedio'] - peor_tienda['Promedio']
print(f"\n📈 DIFERENCIA: {diferencia:.2f} puntos")

# Mostrar tabla comparativa completa
print(f"\n📋 TABLA COMPARATIVA COMPLETA")
print("="*45)
print(df_comparativo.round(2))
```

##**Gráfico de Satisfacción de los clientes**

```python
 #Configuración del estilo
plt.style.use('default')
sns.set_palette("viridis")

# Crear figura principal
fig = plt.figure(figsize=(20, 12))
fig.suptitle('📈 DASHBOARD COMPARATIVO - SATISFACCIÓN DEL CLIENTE POR TIENDA',
             fontsize=18, fontweight='bold', y=0.98)

# Definir el layout del dashboard
gs = fig.add_gridspec(2, 3, hspace=0.3, wspace=0.3)

# GRÁFICO : TERMÓMETRO DE SATISFACCIÓN
ax2 = fig.add_subplot(gs[0, 1])
df_ordenado = df_comparativo.sort_values('Promedio')

# Crear barras horizontales con colores según desempeño
colores_barras = []
for promedio in df_ordenado['Promedio']:
    if promedio >= 4.5:
        colores_barras.append('#2E8B57')  # Verde - Excelente
    elif promedio >= 4.0:
        colores_barras.append('#7CB342')  # Verde claro - Bueno
    elif promedio >= 3.0:
        colores_barras.append('#FFA726')  # Naranja - Regular
    else:
        colores_barras.append('#EF5350')  # Rojo - Malo

bars = ax2.barh(range(len(df_ordenado)), df_ordenado['Promedio'], color=colores_barras, alpha=0.8)

# Añadir etiquetas
for i, (bar, row) in enumerate(zip(bars, df_ordenado.itertuples())):
    ax2.text(bar.get_width() + 0.1, bar.get_y() + bar.get_height()/2,
             f'{row.Promedio:.2f}', va='center', fontweight='bold', fontsize=11)


ax2.set_yticks(range(len(df_ordenado)))
ax2.set_yticklabels([f"{row.Tienda}" for row in df_ordenado.itertuples()], fontweight='bold')
ax2.set_xlim(0, 5.2)
ax2.set_xlabel('Calificación Promedio')
ax2.set_title('🌡️ TERMÓMETRO DE SATISFACCIÓN\n(Ordenado por Desempeño)', fontweight='bold')
ax2.grid(axis='x', alpha=0.3)


```

# 4. Productos más y menos vendidos

```python
# En este punto se busca detallar en cada una de las tiendas los productos que más y menos se venden.

# Función para analizar cada tienda
for nombre, df in tiendas_data:
    top_productos = df['Producto'].value_counts().head(1)
    bottom_productos = df['Producto'].value_counts().tail(1)
    print(f"\nTienda: {nombre}")
    print(f"Producto Más vendido: {top_productos.index[0]}")
    print(f"Cantidad: {top_productos.iloc[0]}")
    print(f"Producto menos vendido: {bottom_productos.index[0]}")
    print(f"Cantidad: {bottom_productos.iloc[0]}")
    print("=" * 50)

```

```python
# se crean listas con la data anterior para poder ver graficamente los productos más y menos vendidos en cada tienda.

# Preparar datos para el gráfico
tiendas = []
productos_mas_vendidos = []
cantidades_mas = []
productos_menos_vendidos = []
cantidades_menos = []

# Ejecutar tu análisis y recolectar datos
for nombre, df in tiendas_data:
    top_productos = df['Producto'].value_counts().head(1)
    bottom_productos = df['Producto'].value_counts().tail(1)

    tiendas.append(nombre)
    productos_mas_vendidos.append(top_productos.index[0])
    cantidades_mas.append(top_productos.iloc[0])
    productos_menos_vendidos.append(bottom_productos.index[0])
    cantidades_menos.append(bottom_productos.iloc[0])

# Crear el gráfico
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Gráfico de productos más vendidos
bars1 = ax1.bar(tiendas, cantidades_mas, color=['#2E86AB', '#A23B72', '#F18F01', '#C73E1D'])
ax1.set_title('Productos Más Vendidos por Tienda', fontsize=14, fontweight='bold')
ax1.set_ylabel('Cantidad Vendida')
ax1.tick_params(axis='x', rotation=45)

# Añadir etiquetas con nombres de productos en las barras
for i, (bar, producto) in enumerate(zip(bars1, productos_mas_vendidos)):
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2., height + 0.1,
             f'{producto}\n({height})', ha='center', va='bottom', fontweight='bold')

# Gráfico de productos menos vendidos
bars2 = ax2.bar(tiendas, cantidades_menos, color=['#2E86AB', '#A23B72', '#F18F01', '#C73E1D'])
ax2.set_title('Productos Menos Vendidos por Tienda', fontsize=14, fontweight='bold')
ax2.set_ylabel('Cantidad Vendida')
ax2.tick_params(axis='x', rotation=45)

# Añadir etiquetas con nombres de productos en las barras
for i, (bar, producto) in enumerate(zip(bars2, productos_menos_vendidos)):
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height + 0.1,
             f'{producto}\n({height})', ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.show()
```
