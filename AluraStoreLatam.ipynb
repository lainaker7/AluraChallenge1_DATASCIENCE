# 1. Análisis de la Facturación

```python
# Se empieza a revisar como han sido las ventas según los precios de los artículos que vende cada tienda
# para definimos la variable de cada tienda y sumamos sus ingresos de acuerdo a los precios.


ventas1 = tienda['Precio'].sum()
ventas2 = tienda2['Precio'].sum()
ventas3 = tienda3['Precio'].sum()
ventas4 = tienda4['Precio'].sum()


```

```python
print(f"💵 Tienda 1: ${ventas1:,.0f} pesos")
print(f"💵 Tienda 2: ${ventas2:,.0f} pesos")
print(f"💵 Tienda 3: ${ventas3:,.0f} pesos")
print(f"💵 Tienda 4: ${ventas4:,.0f} pesos")
```

```python
#Ahora hacemos una lista de todas esas ventas para poder comparar los mejores y peores ingresos de cada tienda.
ventas_totales = [ventas1, ventas2, ventas3, ventas4]
mejor_venta = max(ventas_totales)
peor_venta = min(ventas_totales)

 #Extraemos de los datos cual fue el mejor de los ingresos.
mejor_tienda = ventas_totales.index(mejor_venta) + 1
peor_tienda = ventas_totales.index(peor_venta) + 1
print(f'\n💰 Mejor venta: Tienda {mejor_tienda} tuvo el mayor ingreso')
print(f'⚠️ Peor venta: Tienda {peor_tienda} tuvo el menor ingreso')
```

```python
#Ahora buscamos graficar el resultado para tener la comparativa según el primer analisis realizado.
import matplotlib.pyplot as plt
```

```python

# Datos
ingresos = [ventas1, ventas2, ventas3, ventas4]
tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']
colores = ['#8BB657', '#337B9F', '#9AD1D0', '#B24223']

plt.figure(figsize=(10, 8))
plt.pie(ingresos,
        labels=tiendas,
        colors=colores,
        autopct='%1.1f%%',
        startangle=90,
        explode=(0.05, 0.05, 0.05, 0.05))  # Separar las rebanadas

plt.title('🚀 Ventas por tienda', fontsize=16, fontweight='bold')
plt.axis('equal')  # Para que sea un círculo perfecto
plt.show()

 # 💾 Guardar el gráfico como imagen PNG
plt.tight_layout()
plt.savefig('grafico1_ingresos_tiendas.png', dpi=300, bbox_inches='tight')

# Del análisis se puede verificar que de las ventas totales la tienda 1 tiene el mayor ingreso siendo el 26.1% del total mientras
# que la tienda 4 solo representa un 23.6% de las ventas globales.
```
